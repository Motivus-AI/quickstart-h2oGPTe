```python
from collections import Counter
from pathlib import Path
import json
import sys

ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
tickets_path = ROOT / "data" / "tickets" / "synthetic_tickets.jsonl"
kb_root = ROOT / "data" / "knowledge_base"
sys.path.insert(0, str(ROOT / "packages" / "riot_helpdesk_agent"))

tickets = [json.loads(line) for line in tickets_path.read_text().splitlines() if line.strip()]
print(f"tickets: {len(tickets)}")
print(Counter(t["routing_branch"] for t in tickets))
print(Counter(t["resolution"]["action"] for t in tickets))
```

```python
articles = list(kb_root.rglob("*.md"))
print(f"artículos locales: {len(articles)}")
for path in sorted(articles):
    print(f" - {path.relative_to(kb_root)}")
```
